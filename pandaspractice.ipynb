#1. Create a DataFrame named student_df from a dictionary containing the columns Name, Age, and City for four students.

```python
data ={ 'Name': ['Madhav','Nishant','Charli','Arjun'],
       'Age': [20,22,21,23],
        'City': ['Delhi','Mumbai','Agra','Kanpur'] }
student_df =pd.DataFrame(data)
```

```python
print(student_df)
```

```python
print(student_df.columns)
```

```python
print(student_df['Name'])
```

```python
print(student_df['City'])
```

```python
new_df = student_df[['Name', 'City']]
print(new_df)
```

```python
book_df = pd.DataFrame ({
    'Title': ['The Great Gatsby', 'To Kill a Mockingbird', '1984', 'Pride and Prejudice'],
    'Author': ['F. Scott Fitzgerald', 'Harper Lee', 'George Orwell', 'Jane Austen'],
})
```

```python
book_df['Publisher'] = 'Penguin'
print(book_df)
```

```python
book_df['SerialNo']= np.arange(4)
print(book_df)
```

```python
book_df = book_df.drop('SerialNo', axis=1)
print(book_df)
```
